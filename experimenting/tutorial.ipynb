**Interpreting the plots**

Based off of the data, it appears that emeralds are a fixed value asset, with their spread staying around a value of $10k$

Looking at the tomatoes graph, it is more of an unstable asset, whose price is more volatile, resembling a random walk. Next steps would be to test for a unit root in the tomato data to confirm this and test for any possible relationship/cointegration with emeralds

```python
from statsmodels.tsa.stattools import adfuller
# Function to organize ADF test results
def adf_test(data):
    keys = ['Test Statistic','p-value','# of Lags','# of Obs']
    values = adfuller(data)
    test = pd.DataFrame.from_dict(dict(zip(keys,values[0:4])),
                                  orient='index',columns=[data.name])
    return test
```

```python
cols = ['bid_price_1', 'bid_price_2', 'ask_price_1', 'ask_price_2', 'mid_price']
```

```python
dummy = tomatoes_01.drop(columns=[col for col in tomatoes_01.columns if col not in cols])
temp = []
for column in dummy.columns:
    test = adf_test(dummy[column].dropna())
    temp.append(test)
results = pd.concat(temp, axis=1)
display(temp)
```

```python
dummy = tomatoes_02.drop(columns=[col for col in tomatoes_01.columns if col not in cols])
temp = []
for column in dummy.columns:
    test = adf_test(dummy[column].dropna())
    temp.append(test)
results = pd.concat(temp, axis=1)
display(temp)
```

Given the p-values, we fail to reject the null hypothesis that the tomato data has a unit root/is non-stationary. Will now try differencing the data to confirm the noise is stationary

```python
diff_tomatoes_01 = pd.DataFrame()
diff_tomatoes_01['diff_ask_1'] = tomatoes_01['ask_price_1'] - tomatoes_01['ask_price_1'].shift(1)
diff_tomatoes_01['diff_ask_2'] = tomatoes_01['ask_price_2'] - tomatoes_01['ask_price_2'].shift(1)
diff_tomatoes_01['diff_bid_1'] = tomatoes_01['bid_price_1'] - tomatoes_01['bid_price_1'].shift(1)
diff_tomatoes_01['diff_bid_2'] = tomatoes_01['bid_price_2'] - tomatoes_01['bid_price_2'].shift(1)
diff_tomatoes_01['diff_mid'] = tomatoes_01['mid_price'] - tomatoes_01['mid_price'].shift(1)
diff_tomatoes_01
```

```python
temp = []
for column in diff_tomatoes_01.columns:
    test = adf_test(diff_tomatoes_01[column].dropna())
    temp.append(test)
results = pd.concat(temp, axis=1)
display(temp)
```
